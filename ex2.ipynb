```python
#Monthly or year expense
print("Monthly or Yearly Expense")
expenses=[]
while True:
    print(f"1.ADD EXPENSE\n 2.VIEW EXPENSES \n 3.TOTAL EXPENSE \n 4.SAVE AND EXIT\n")
    choice=int(input("enter your choice:"))
    if choice==1:
        amount=int(input("enter the amount"))
        category=input("enter the category")
        note=input("enter the notes")
        expense={
            'amount':amount,
            'category':category,
            'note':note}
        expenses.append(expense)
        print("successfully added")
    if choice==2:
        for exp in expenses:
            print(exp)
        print("viewed successfully")
            #print(f"Amount:{exp[amount]}\n Category:{exp[category]}\n Note:{exp[note]}")
    if choice==3:
        total=sum(exp["amount"] for exp in expenses)
        print(f"total amount:",total)

    if choice==4:
        #save file expense.txt
        with open("expense.txt",'w')as file:
            for exp in expenses:
                file.write(str(exp)+"\n")
        print("updated in the file")
        break
    
```

```python
#save the expense in csv file
expenses=[]
while True:
    print(f"1.ADD EXPENSE\n 2.VIEW EXPENSES \n 3.TOTAL EXPENSE \n 4.SAVE AND EXIT\n")
    choice=int(input("enter your choice:"))
    if choice==1:
        eb_bill=int(input("this month eb_bill"))
        phone_bill=int(input("phone_bill payment:"))
        balance=int(input("ur balance:"))
        value={
            'eb_bill':eb_bill,
            'phone_bill':phone_bill,
            "balance":balance
        }
        expenses.append(value)
        print("added successfully")
    if choice==2:
        for val in expenses:
            print(val)
    if choice==3:
        total_ebbill=sum(val['eb_bill'] for val in expenses)
        total_ph_bill=sum(val['phone_bill']for val in expenses)
        total_bal=sum(val['balance'] for val in expenses)
        print(f"\neb_bill:{total_ebbill}\n phone_bill:{total_ph_bill}\n balance:{total_bal}")
    if choice==4:
        with open("expense.csv",'w',newline="")as file:
            file.write(str(expenses)+"\n")
            break
        
```
